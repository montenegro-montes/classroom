## Implementación de IDA*

```python
class Node:
    def __init__(self, state, parent=None, cost=0):
        self.state = state
        self.parent = parent
        self.cost = cost

    def __str__(self):
        return f"Node(state={self.state}, parent={self.parent}, cost={self.cost})"

    def __lt__(self, other):
        return self.cost < other.cost

def ida_star(problem):
    bound = problem.f(Node(problem.initial_state))
    
    path = [Node(problem.initial_state)]
    while True:
        result, new_bound = depth_limited_search(problem, bound, path)
        if result == 'FOUND':
            return path
        if result == float('inf'):
            return 'No se encontró camino'
        bound = new_bound
        print("\n\n new bound", bound)

def depth_limited_search(problem, bound, path,):

    node = path[-1]
   
    f = problem.f(node)    
    print("NODE", node, "f",f)
    
    if problem.f(node) > bound:
        print("\t Poda NODE", node, "f",f)
        return f, 0
    
    if problem.is_goal(node.state):
        return 'FOUND', 0

    min_val = float('inf')
    for child_state, cost in problem.expand(node.state):
        if child_state not in [n.state for n in path]:
            child_node = Node(child_state, node, node.cost + cost)  
            path.append(child_node)
            result, _ = depth_limited_search(problem, bound, path)
            if result == 'FOUND':
                return 'FOUND', 0
            if result < min_val:
                min_val = result
            path.pop()
    return min_val, min_val
```

## Problema de Rumania

```python
class MapaRumaniaProblem:
    def __init__(self, graph, initial_state, goal_state,f):
        self.graph = graph
        self.initial_state = initial_state
        self.goal_state = goal_state
        self.f = f

    def is_goal(self, state):
        return state == self.goal_state

    def expand(self, state):
        return [(child_state, cost) for child_state, cost in self.graph[state].items()]

# Definimos el grafo de Rumania
graph = {
    'Arad': {'Zerind': 75, 'Sibiu': 140, 'Timisoara': 118},
    'Zerind': {'Arad': 75, 'Oradea': 71},
    'Oradea': {'Zerind': 71, 'Sibiu': 151},
    'Sibiu': {'Arad': 140, 'Oradea': 151, 'Fagaras': 99, 'Rimnicu Vilcea': 80},
    'Timisoara': {'Arad': 118, 'Lugoj': 111},
    'Lugoj': {'Timisoara': 111, 'Mehadia': 70},
    'Mehadia': {'Lugoj': 70, 'Drobeta': 75},
    'Drobeta': {'Mehadia': 75, 'Craiova': 120},
    'Craiova': {'Drobeta': 120, 'Rimnicu Vilcea': 146, 'Pitesti': 138},
    'Rimnicu Vilcea': {'Sibiu': 80, 'Craiova': 146, 'Pitesti': 97},
    'Fagaras': {'Sibiu': 99, 'Bucharest': 211},
    'Pitesti': {'Rimnicu Vilcea': 97, 'Craiova': 138, 'Bucharest': 101},
    'Bucharest': {'Fagaras': 211, 'Pitesti': 101}
}

def f(node):
    return node.cost + heuristic(node.state)

# Definimos la función heurística
def heuristic(city):
    distances_to_bucharest = {
        'Arad': 366,
        'Bucharest': 0,
        'Craiova': 160,
        'Drobeta': 242,
        'Eforie':161,
        'Fagaras': 176,
        'Giurgiu':77,
        'Hirsova':151,
        'Iasi':226,
        'Lugoj': 244,
        'Mehadia': 241,
        'Neamt':234,
        'Oradea': 380,
        'Pitesti':100,
        'Rimnicu Vilcea': 193,
        'Sibiu': 253,
        'Timisoara': 329,
        'Urziceni':80,
        'Vaslui':199,
        'Zerind': 374
    }
    return distances_to_bucharest[city]



problem = MapaRumaniaProblem(graph, 'Arad', 'Bucharest',f)
# Resolvemos el problema utilizando IDA*
solution_path = ida_star(problem)
print("\n\nCamino más corto encontrado utilizando IDA*:")
for node in solution_path:
    print(node.state)

```
