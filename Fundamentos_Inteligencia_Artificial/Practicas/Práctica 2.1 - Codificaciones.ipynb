# Cantor Pairing Function

La **función de emparejamiento de Cantor** permite codificar varios números naturales en **un único número natural** de forma biyectiva.

Esto significa que existe una correspondencia **uno a uno** entre:

- pares (o tuplas) de números naturales  
- un único número natural  

Esta función se utiliza en **teoría de la computabilidad**, **Gödelización** y **codificación de programas**.

---

### Cantor2_1

La función de cantor2_1 para dos variables es:

$\delta_2^1(x,y) = \frac{(x+y)(x+y+1)}{2} + y$

Esta función codifica un **par de números naturales** en uno solo.

### Cantor3_1

La función de cantor3_1 para tres variables es:

\\ ToDO

Esta función codifica un **tres de números naturales** en uno solo.

### Cantork_1

La función de cantork_1 para n variables es:

\\ ToDO

Esta función codifica un **n  números naturales** en uno solo.


```python
def cantor2_1(x,y):
    sol = (x+y)*(x+y+1)//2 + y

    return sol

def cantor3_1(x,y,z):
   sol = cantor2_1(x,cantor2_1(y,z))

   return sol

def cantor_k_1(*args):
    
    # caso base
    if len(args) == 1:
        return args[0]

    # caso recursivo
    return cantor2_1(args[0], cantor_k_1(*args[1:]))
```

# Gödel Encoding

La **codificación de Gödel** es un método que permite representar **secuencias de números naturales como un único número natural**.

Esta idea es fundamental en **teoría de la computabilidad** porque permite:

- codificar programas
- codificar fórmulas lógicas
- enumerar estructuras matemáticas

### Definición

Sea \(x = (x_1,x_2,\dots,x_m)\) una secuencia de números naturales.

La función de Gödel se define como:

$$
god(x_1,x_2,\dots,x_m) =
\rho_2^1(m-1,\rho_1^1(x))
+ 1
$$

donde:

- \(m\) es la **longitud del vector**
- \(\rho_1^1(x)\) es la **codificación del vector usando Cantor**
- \(\rho_2^1\) es la **función de Cantor para pares**

---

### Casos

#### Vector vacío

$$
god() = 0
$$

#### Vector de un elemento

$$
god(x_1) = \delta_2^1(0,x_1) + 1
$$

#### Vector de varios elementos

$$
god(x_1,x_2,\dots,x_m)
=
\delta_2^1(m-1,\delta_1^1(x_1,x_2,\dots,x_m)) + 1
$$

---

### Ejemplo

Para el vector

$$
(1,2,3)
$$

primero codificamos el vector usando Cantor:

$$
\delta_1^1(1,2,3) =
\delta_2^1(1,\delta_2^1(2,3))
$$

Después añadimos la longitud:

$$
god(1,2,3) =
\delta_2^1(2,\delta_1^1(1,2,3)) + 1
$$

---

```python
def god(*args):

    if len(args) == 0:
        return 0

    L = len(args)

    contenido = cantor_k_1(*args)

    return cantor2_1(L-1, contenido) + 1

```

```python
ej1 = god(0)
ej2 = god(1)
ej3 = god(2)

ej4 = god(0,0)
ej5 = god(1,0)
ej6 = god(0,0,0)

print(ej1)
print(ej2)
print(ej3)
print(ej4)
print(ej5)
print(ej6)
```

# Parsing de programas WHILE

Para convertir un programa WHILE escrito como texto en la representación interna `S`, utilizamos dos funciones:

- `parse_line`: convierte **una línea del programa** en su representación vectorial.
- `parse_program`: convierte **todo el programa** en una lista de instrucciones.

---

### Representación interna de instrucciones

Cada instrucción se representa como un vector:

| Operación | Representación |
|-----------|---------------|
| `Xi := 0` | `[0, i]` |
| `Xi := Xj` | `[1, i, j]` |
| `Xi := Xj + 1` | `[2, i, j]` |
| `Xi := Xj - 1` | `[3, i, j]` |
| `while Xi != 0 do S od` | `[4, i, S]` |

---

### Función `parse_line`

Esta función analiza una **línea de código WHILE** y la convierte en su representación interna.

### Función `parse_program`

Esta función analiza un programa completo y genera su representación S.

```python
def parse_line(line):

    line = line.strip()

    # Xi := 0
    if ":= 0" in line:
        i = int(line.split("X")[1].split()[0])
        return [0, i]

    # Xi := Xj + 1
    if "+ 1" in line:
        parts = line.replace("X","").replace(":=","").split()
        i = int(parts[0])
        j = int(parts[1])
        return [2, i, j]

    # Xi := Xj - 1
    if "- 1" in line:
        parts = line.replace("X","").replace(":=","").split()
        i = int(parts[0])
        j = int(parts[1])
        return [3, i, j]

    # Xi := Xj
    if ":=" in line:
        parts = line.replace("X","").replace(":=","").split()
        i = int(parts[0])
        j = int(parts[1])
        return [1, i, j]

    return None

####
# parse_program
###

def parse_program(lines):

    S = []
    i = 0

    while i < len(lines):

        line = lines[i].strip()

        # while
        if line.startswith("while"):

            var = int(line.split("X")[1].split()[0])

            cuerpo = []

            i += 1
            while lines[i].strip() != "od":
                cuerpo.append(parse_line(lines[i]))
                i += 1

            S.append([4, var, cuerpo])

        else:
            S.append(parse_line(line))

        i += 1

    return S
```

```python
programa = [
"while (X2 != 0) do",
"X1 := X1 + 1",
"od"
]

S = parse_program(programa)

print(S)
```

## Función `while_program(S)`

La función `while_program` convierte la representación interna de un programa WHILE en una **versión legible en pseudocódigo**.

El programa se representa mediante una lista `S` donde cada instrucción es un vector.

La función `print_Q` imprime la terna del programa WHILE (n,p,s) siendo s:  una **versión legible en pseudocódigo**.



```python
def while_program(S, nivel=0):
    tab = "    " * nivel

    program = ""

    for sent in S:
        op = sent[0]

        if op == 0:
            i = sent[1]
            program += f"{tab}X{i} := 0\n"

        elif op == 1:
            i, j = sent[1], sent[2]
            program += f"{tab}X{i} := X{j}\n"

        elif op == 2:
            i, j = sent[1], sent[2]
            program += f"{tab}X{i} := X{j} + 1\n"
          
        elif op == 3:
            i, j = sent[1], sent[2]
            program += f"{tab}X{i} := X{j} - 1\n"
          
        elif op == 4:
            i = sent[1]
            cuerpo = sent[2]
            program += f"{tab}while (X{i} != 0) do\n"
            program += while_program(cuerpo, nivel + 1)
            program += f"{tab}od\n"

    return program

####### Imprime Q
def print_Q(Q):
    n, p, s = Q
    print(f"Q = ({n}, {p}, s)")
    print("s:")
    print(while_program(s))
```

# Generación aleatoria de programas WHILE

Para realizar experimentos con programas WHILE podemos generar programas **aleatorios**.  
La idea es construir programas utilizando las instrucciones del lenguaje WHILE de forma automática.

---

## Función `genera_instruccion`

La función `genera_instruccion` genera **una instrucción aleatoria** del lenguaje WHILE.

Parámetros:

- `num_vars`: número máximo de variables disponibles.
- `profundidad_max`: profundidad máxima permitida para bucles `while`.
- `profundidad_actual`: profundidad actual del programa.

Si se alcanza la profundidad máxima, no se generan más instrucciones `while` para evitar bucles demasiado profundos.

## Función `genera_lenguaje`

Esta función genera un programa WHILE completo como una lista de instrucciones.

## Función `genera_Q`

Un programa WHILE completo se representa como:

Q=(n,p,s)

donde:

- n es el número de variables de entrada
- p es el número total de variables
- s es el programa WHILE

La función genera_Q genera una instancia aleatoria de esta estructura.

```python
import random

def genera_instruccion(num_vars, profundidad_max, profundidad_actual=0):
    if profundidad_actual >= profundidad_max:
        op = random.randint(0, 3)
    else:
        op = random.randint(0, 4)

    if op == 0:   # Xi := 0
        i = random.randint(1, num_vars)
        return [0, i]

    elif op == 1: # Xi := Xj
        i = random.randint(1, num_vars)
        j = random.randint(1, num_vars)
        return [1, i, j]

    elif op == 2: # Xi := Xj + 1
        i = random.randint(1, num_vars)
        j = random.randint(1, num_vars)
        return [2, i, j]

    elif op == 3: # Xi := Xj - 1
        i = random.randint(1, num_vars)
        j = random.randint(1, num_vars)
        return [3, i, j]

    elif op == 4: # while Xi != 0 do S od
        i = random.randint(1, num_vars)
        cuerpo = genera_lenguaje(
            num_vars=num_vars,
            num_instrucciones=random.randint(1, 3),
            profundidad_max=profundidad_max,
            profundidad_actual=profundidad_actual + 1
        )
        return [4, i, cuerpo]


def genera_lenguaje(num_vars=3, num_instrucciones=3, profundidad_max=2, profundidad_actual=0):
    programa = []

    for _ in range(num_instrucciones):
        sent = genera_instruccion(num_vars, profundidad_max, profundidad_actual)
        programa.append(sent)

    return programa


def genera_Q(n_max=3, p_max=5, instrucciones_max=4, profundidad_max=2):
    n = random.randint(1, n_max)
    p = random.randint(n, p_max)
    s = genera_lenguaje(
        num_vars=p,
        num_instrucciones=random.randint(1, instrucciones_max),
        profundidad_max=profundidad_max
    )
    return (n, p, s)
```

```python
Q = genera_Q()

print(Q)
print()
print_Q(Q)
```

# Función `codi`

La función `codi` codifica **una instrucción del lenguaje WHILE** como un número natural.

Esta codificación se utiliza para construir posteriormente la **codificación de programas completos** mediante la función `Codi`.

La idea consiste en asociar a cada tipo de instrucción un número distinto usando la función de Cantor.


```python

def codi(sent):
    op = sent[0]

    if op == 0:                      # Xi := 0
        i = sent[1]
        return 5 * (i - 1)

    elif op == 1:                    # Xi := Xj
        i, j = sent[1], sent[2]
        return 5 * cantor2_1(i - 1, j - 1) + 1

    elif op == 2:                    # Xi := Xj + 1
        i, j = sent[1], sent[2]
        return 5 * cantor2_1(i - 1, j - 1) + 2

    elif op == 3:                    # Xi := Xj - 1
        i, j = sent[1], sent[2]
        return 5 * cantor2_1(i - 1, j - 1) + 3

    elif op == 4:                    # while Xi != 0 do s1 od
        i = sent[1]
        s1 = sent[2]
        return 5 * cantor2_1(i - 1, Codi(s1)) + 4

    else:
        raise ValueError("Operacion no soportada")
```

# Función `Codi`

La función `Codi` codifica **un programa WHILE completo** como un único número natural.

Para ello se utilizan dos pasos:

1. Codificar **cada instrucción** del programa mediante la función `codi`.
2. Codificar la **secuencia de instrucciones** usando la función de Gödel `god`.

---

```python

def Codi(programa):

    # programa vacío
    if len(programa) == 0:
        return -1

    codigos = []

    # codificamos cada instrucción
    for sent in programa:
        codigos.append(codi(sent))

    # aplicamos la codificación de Gödel
    resultado = god(*codigos)

    return resultado - 1
```

# Función `CODI`

La función `CODI` permite codificar **un programa WHILE completo** como un número natural.  
Un programa WHILE se representa como una tupla:

\[
Q = (n, p, s)
\]

donde:

- \(n\) es el número de **variables de entrada**
- \(p\) es el número total de **variables del programa**
- \(s\) es el **programa WHILE** (lista de instrucciones)

La función `CODI` transforma esta información en un único número natural usando la función de Cantor para tres variables.

---

# Definición

La codificación del programa se define como:

$CODI(Q) = \delta_3^1(n,\; p - \max(n,k),\; Codi(s))$

donde:

- \(k = \max\{m \mid X_m \text{ aparece en } s\}\)
- `Codi(s)` es la codificación del programa
- \(\rho_3^1\) es la función de Cantor para tres variables.

El término \(p - \max(n,k)\) asegura que el número de variables del programa sea suficiente para representar todas las variables utilizadas.


```python
def max_variable(S):

    k = 0

    for sent in S:

        op = sent[0]

        if op == 0:
            k = max(k, sent[1])

        elif op in [1,2,3]:
            k = max(k, sent[1], sent[2])

        elif op == 4:
            k = max(k, sent[1], max_variable(sent[2]))

    return k

######


def CODI(Q):

    n = Q[0]
    p = Q[1]
    s = Q[2]

    k = max_variable(s)

    return cantor3_1(n, p - max(n,k), Codi(s))
```

Ejecutar un programa while, en **fase experimental**

```python
def execute(S, X, max_steps=1000, steps=None):

    if steps is None:
        steps = [0]   # contador compartido

    for sent in S:

        if steps[0] >= max_steps:
            print("Max steps reached")
            return None

        op = sent[0]
        steps[0] += 1

        if op == 0:                     # Xi := 0
            i = sent[1]
            X[i] = 0 
            X[0] +=1

        elif op == 1:                   # Xi := Xj
            i, j = sent[1], sent[2]
            X[i] = X[j]
            X[0] +=1

        elif op == 2:                   # Xi := Xj + 1
            i, j = sent[1], sent[2]
            X[i] = X[j] + 1
            X[0] +=1

        elif op == 3:                   # Xi := Xj - 1
            i, j = sent[1], sent[2]
            X[i] = max(0, X[j] - 1)
            X[0] +=1
            
        elif op == 4:                   # while Xi != 0 do S od
            i      = sent[1]
            cuerpo = sent[2]
          
            while X[i] != 0:
            
                if steps[0] >= max_steps:
                    print("Max steps reached")
                    return None

                execute(cuerpo, X, max_steps, steps)
            
    return X

#########
####
###
def configuracion_inicial(n,p):
    X = [1] + [random.randint(1,5) for _ in range(n)] + [0]*(p-n)
    return X

#########
####
###

def F_Q(Q):
    #print(Q)
    #print()
    print_Q(Q)

    n=Q[0]
    p=Q[1]
    #print(n,p)
    S = Q[2]

    C_ini   =  configuracion_inicial(n,p)
    print(f"Configuracion inicial: {C_ini }")
    C_final =  execute(S,C_ini )
    print(f"Configuracion final: {C_final }")
    
```
