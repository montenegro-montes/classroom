### Algoritmo de búsqueda binaria.

Cada paso el algoritmo busca el elemento intermedio del vector. 
- Si el valor coincide con el elemento intermedio devuelve la posición
- Si el valor es menor que el elemento intermedio nos quedamos con la mitad del vector de la izquierda
- En caso contrario nos quedamos con la parte derecha del vector.


```python
def busqueda_binaria(data, value):
    n = len(data)
    izq = 0
    dch = n - 1
    
    while izq <= dch:
        mitad = (izq + dch) // 2
        
        if value == data[mitad]:
            return mitad
        elif value < data[mitad]:
            dch = mitad - 1
        else:
            izq = mitad + 1
    return -1

```

#### Vamos añadir algunos elementos al algortimo original para observar cuántas iteraciones son realizadas en cada uno de los casos. Con el número de iteraciones podemos observar que el peor de los casos hace que el algoritmo sea catalogado como O(log n) siendo n la longitud del vector.

```python
def busqueda_binaria_pasos(data, value):
    n = len(data)
    izq = 0
    dch = n - 1
    i = 1 # Externo algoritmo
    
    while izq <= dch:
        mitad = (izq + dch) // 2
        print('\t indice mitad ',mitad, ' data[mitad] ',data[mitad])
        
        if value == data[mitad]:
            print('   # Iteraciones:',i) # Externo algoritmo
            return mitad
        elif value < data[mitad]:
            dch = mitad - 1
        else:
            izq = mitad + 1
            
        i+=1 # Externo algoritmo
    return -1
```

## Ejemplo de O(n)

Al igual que el caso anterior n determina la longitud del vector. 
El algoritmo de búsqueda lineal comenzando por el primer elemento tiene que iterar sobre todo el vector hasta encontrar el valor que busca.

Una ventaja sobre el algoritmo de búsqueda binaria es que no obliga a que el vector esté ordenado.


```python
def busqueda_lineal(datos, valor):
    for indice in range(len(datos)):
        if valor == datos[indice]:
            return indice
    return -1    
```

```python
%timeit busqueda_lineal(numeros_aleatorios, mitad)
%timeit busqueda_lineal(numeros_aleatorios, primero)
%timeit busqueda_lineal(numeros_aleatorios, ultimo)
```

## Ejemplo de O(n^2)

Al igual que los casos anteriores n determina la longitud del vector. 
El algoritmo de ordenamiento burbuja itera por todos los elementos del vector hasta encontrar una posición que sea la adecuada para el valor, con lo que necesita comparar con todos los elementos que sean menor que el valor selecionado.

```python
def ordenamiento_burbuja(data):
    intercambio = True
    while intercambio:
        intercambio = False
        for i in range(len(data)-1):
            if data[i] > data[i+1]:
                data[i], data[i+1] = data[i+1], data[i]
                intercambio = True
        print('\t', data)  # Externo algoritmo
```

#### Vamos a ver el funcionamiento del algoritmo en el peor de los casos. Para ello creamos de forma aleatorio un vector ordenado de forma descendente, que representa el peor de los casos para el algoritmo.

```python
n=10
numeros_aleatorios_Burbuja = random.sample(range(1, n*4), n)  # Genera n números aleatorios del 1 al n*4
numeros_aleatorios_Burbuja.sort(reverse=True)                 # peor caso orden inverso
print(numeros_aleatorios_Burbuja)
ordenamiento_burbuja(numeros_aleatorios_Burbuja)
```

```python
def ordenamiento_burbuja_pasos(data):
    intercambio = True
    cont=0;           # Externo algoritmo
    while intercambio:
        intercambio = False
        for i in range(len(data)-1):
            cont+=1;  # Externo algoritmo
            if data[i] > data[i+1]:
                data[i], data[i+1] = data[i+1], data[i]
                intercambio = True
    return cont   
```

#### Hemos modificado el algoritmo para contabilizar el número de pasos necesarios para realizar el orden según un n de entrada. Este algoritmo nos va a permitir crear un gráfico que muestre el compartamiento del algoritmo según el valor de n. Comparamos la gráfica obtenida con la gráfica de la función n^2

```python
pasos = []

n=10
        
for i in range (0,100,n):
        numeros_aleatorios_Burbuja = random.sample(range(1, i*4), i)  
        numeros_aleatorios_Burbuja.sort(reverse=True) # peor caso
        pasos.append (ordenamiento_burbuja_pasos(numeros_aleatorios_Burbuja))
print(pasos)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0,100,10)
y = pasos

quadratic = x ** 2

plt.rcParams["figure.autolayout"] = True
plt.title("O(n^2)")
plt.plot(x, y, color="red")
plt.plot(x, quadratic, color='blue')
plt.legend(['$Bubble Sort$','$\mathcal{O}(n^2)$', ], loc='best', fontsize=20);

plt.show()
```

## Ejemplo de O(2^n)



```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

serie = []

for i in range (0,11):
    serie.append(fibonacci(i))

print (serie)
#Valor de x[i] es calculado mediante la x[i-1] + x[i-2]
#p.ej x[4] = x[3] + x[2]= 2 + 1 = 3
    
```

```python
pasos = 0

def fibonacci(n):
    global pasos
    if n <= 1:
        pasos+=1
        return n
    pasos+=2
    return fibonacci(n-1) + fibonacci(n-2)

step  = []
index = []

for i in range (0,31):
    pasos = 0
    fibonacci(i)
    #print(i, " Resultado", fibonacci(i))
    #print("\t PASOS: ",pasos)
    step.append(pasos)
    index.append(i)
    


x = np.arange(0,23)
y = 2 ** x

plt.rcParams["figure.autolayout"] = True
plt.title("O(2^n)")
plt.plot(index, step, color="red")
plt.plot(x, y, color='blue')
plt.legend(['$Fibonacci$','$\mathcal{O}(2^n)$', ], loc='best', fontsize=20);
plt.show()
```
