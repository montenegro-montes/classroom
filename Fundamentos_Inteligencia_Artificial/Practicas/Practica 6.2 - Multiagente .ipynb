# Algoritmo MINIMAX

```python
def minimax_search(game, state):
    game.player = game.to_move(state)
    value, move = max_value(game, state)
    return move

def max_value(game, state):
    if game.is_terminal(state):
        utility =  game.utility(state, game.player)
        return utility, None
    v = float("-inf")
    for a in game.actions(state):
        v2, _ = min_value(game, game.result(state, a))
        if v2 > v:
            v, move = v2, a
    return v, move

def min_value(game, state):
    if game.is_terminal(state):
        utility =  game.utility(state, game.player)
        return utility, None
    v = float("inf")
    for a in game.actions(state):
        v2, _ = max_value(game, game.result(state, a))
        if v2 < v:
            v, move = v2, a
    return v, move
```

# Algoritmo AlFA-BETA

```python
def alpha_beta_search(game, state):
    game.player = game.to_move(state)
    alpha = float("-inf")
    beta = float("inf")
    _, move = max_valueAB(game, state, alpha, beta)
    return move

def max_valueAB(game, state, alpha, beta):
    if game.is_terminal(state):
        return game.utility(state, game.to_move(state)), None
    v = float("-inf")
    for action in game.actions(state):
        v2, _ = min_valueAB(game, game.result(state, action), alpha, beta)
        if v2 > v:
            v = v2
            move = action
        if v >= beta:
            return v, move
        alpha = max(alpha, v)
    return v, move

def min_valueAB(game, state, alpha, beta):
    if game.is_terminal(state):
        return game.utility(state, game.to_move(state)), None
    v = float("inf")
    for action in game.actions(state):
        v2, _ = max_valueAB(game, game.result(state, action), alpha, beta)
        if v2 < v:
            v = v2
            move = action
        if v <= alpha:
            return v, move
        beta = min(beta, v)
    return v, move
```

```python
class TicTacToeGame:

    def __init__(self, initial_state=None):
        if initial_state is None:
            self.state = [[' ' for _ in range(3)] for _ in range(3)]
        else:     
            self.state = initial_state
        player = 'X'

    def to_move(self, state):
        count_X = sum(row.count('X') for row in state)
        count_O = sum(row.count('O') for row in state)
        if count_X <= count_O:
            return 'X'
        else:
            return 'O'

    def is_terminal(self, state):
        return self.check_winner(state) or self.is_full(state)

    def utility(self, state, player):
       return 0

    def actions(self, state):
        moves = []
        for i in range(3):
            for j in range(3):
                if state[i][j] == ' ':
                    moves.append((i, j))
        return moves

    def result(self, state, action):
        player = self.to_move(self.state)
        row, col = action
        new_state = [row[:] for row in state]
        new_state[row][col] =  player  #self.current_player
        return new_state

    def check_winner(self, state):
        for i in range(3):
            if state[i][0] == state[i][1] == state[i][2] != ' ':
                return state[i][0]
            if state[0][i] == state[1][i] == state[2][i] != ' ':
                return state[0][i]
        if state[0][0] == state[1][1] == state[2][2] != ' ':
            return state[0][0]
        if state[0][2] == state[1][1] == state[2][0] != ' ':
            return state[0][2]
        return None

    def is_full(self, state):
        for row in state:
            for cell in row:
                if cell == ' ':
                    return False
        return True
    
    def setMove(self,move):
        i = move[0]
        j = move[1]  
        player = self.to_move(self.state)

        self.state[i][j] = player 
        print(self.state)

    def game_over(self):
        winner = self.check_winner(self.state)
        if winner or self.is_full(self.state):
            return True
        return False
    
    def winner(self):    
        return  self.check_winner(self.state)
```
