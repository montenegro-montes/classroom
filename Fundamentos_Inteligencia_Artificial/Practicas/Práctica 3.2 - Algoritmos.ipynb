### AC3 y Revise

Ejecuta los algoritmos 'AC3' y 'Revise'

Verifique su correcto funcionamiento

```python
from collections import deque

def revise(domains, constraints, xi, xj):
    revised = False
    for x in domains[xi][:]:
        satisfies = False
        for y in domains[xj]:
            if constraint_function( x, y):
                satisfies = True
                break
        if not satisfies:
            domains[xi].remove(x)
            revised = True
    return revised

def ac3(domains, constraints):
    queue = deque(constraints)
    while queue:
        xi, xj = queue.popleft()
        if revise(domains, constraints, xi, xj):
            if not domains[xi]:
                return False
            for xk, xl in constraints:
                if xl == xi and xk != xj:
                    queue.append((xk, xi))
    return True

# Ejemplo de uso
variables = ['A', 'B', 'C']
domains = {'A': [1, 2], 'B': [1], 'C': [1, 2, 3]}

def constraint_function( x, y):
    return x!=y 

#A!=B, A!=C, B!=C
constraints = [
    ('A', 'B'), ('B', 'A'),
    ('A', 'C'), ('C', 'A'),
    ('B', 'C'), ('C', 'B')
]

print("Dominios antes de AC3:")
print(domains)


if ac3(domains, constraints):
    print("\nDominios después de AC3:")
    print(domains)
else:
    print("\nNo se encontró solución.")
```

### Heurstica LCV

Verifique el correcto funcionamiento de LCV

```python
def least_constraining_value(variable, domains, assignment, constraints):
    max_remaining_values = float('-inf')
    lcv_value = None
    
    for value in domains[variable]:
        assignment[variable] = value
        remaining_values = 0
        
        print(f"{variable}: {value}")
        for other_variable in domains:
            if other_variable != variable and assignment[other_variable] is None:
                for other_value in domains[other_variable]:
                    print(f"\t{other_variable}: {other_value}")
                    if constraints(value, other_value):
                        remaining_values += 1

        print(f"{variable}: Remain {remaining_values}")
        if remaining_values > max_remaining_values:
            max_remaining_values = remaining_values
            lcv_value = value

    assignment[variable] = None   # restaurar
    return lcv_value

# Ejemplo de uso:
variables = ['A', 'B', 'C']
domains = {'A': [1, 2, 3], 'B': [1,3], 'C': [2,3]}
assignment = {'A': None, 'B': None, 'C': None}
neighbors = {
    'A': ['B', 'C'],
    'B': ['A'],
    'C': ['A']
}

def constraints( x, y):
    return x != y

variable_a = 'A'  # Por ejemplo, supongamos que queremos encontrar el valor menos restrictivo para la variable A
lcv_value = least_constraining_value(variable_a, domains, assignment, constraints)
print("El valor menos restrictivo para la variable", variable_a, "es:", lcv_value)
```

### Heurstica MRV

Verifique el correcto funcionamiento de MRV

```python
def minimum_remaining_values(variables, domains, assignment):
    min_remaining_values = float('inf')
    mrv_variable = None
    
    for variable in variables:
        if assignment[variable] is None:  # Si la variable no está asignada
            remaining_values = len(domains[variable])

            if remaining_values == 0:
                return None  
            
            if remaining_values < min_remaining_values:
                min_remaining_values = remaining_values
                mrv_variable = variable
    
    return mrv_variable

# Ejemplo de uso
variables = ['A', 'B', 'C']
domains = {'A': [1, 2, 3], 'B': [1, 2], 'C': [1]}
assignment = {'A': None, 'B': None, 'C': 1}

mrv_variable = minimum_remaining_values(variables, domains, assignment)
print("La variable con el mínimo número de valores restantes es:", mrv_variable)
```

###  Degree Heurstic

Verifique el correcto funcionamiento de la Heurística de grado

```python
def degree_heuristic(variables, neighbors, assignment):
    max_degree = float('-inf')
    selected_variable = None

    for var in variables:
        if assignment[var] is None:
            # contar vecinos no asignados
            degree = sum(
                1 for neighbor in neighbors[var]
                if assignment[neighbor] is None
            )

            if degree > max_degree:
                max_degree = degree
                selected_variable = var

    return selected_variable

# Ejemplo de uso
variables = ['A', 'B', 'C', 'D']

neighbors = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

assignment = {
    'A': None,
    'B': None,
    'C': 1,
    'D': None
}

deg_variable = degree_heuristic(variables, neighbors, assignment)
print("La variable con el mayor grado es:", deg_variable)
```
